```python
import os
import pandas as pd

class_mapping = {
    0: 'person',  # PERSON
    1: 'face',    # FACE
    2: 'cellphone',  # MOBILE
    3: 'hand'     # HAND
}

# txt_folder = '/home/ajeet/codework/datasets/Training_Validation_Dataset_2024_Ajeet-20241107T105545Z-001/Training_Validation_Dataset_2024_Ajeet/test_2021/2021/mobile_output'
# output_csv = 'data/test_mobile_output_annotations.csv'

# txt_folder = '/home/ajeet/codework/datasets/Training_Validation_Dataset_2024_Ajeet-20241107T105545Z-001/Training_Validation_Dataset_2024_Ajeet/test_2021/2021/test/test/aug'
# output_csv = 'data/test_test_aug_annotations.csv'

txt_folder = '/home/ajeet/codework/datasets/Training_Validation_Dataset_2024_Ajeet-20241107T105545Z-001/Training_Validation_Dataset_2024_Ajeet/test_2021/2021/test/test/nov_dec'
output_csv = 'data/test_test_nov_dec_annotations.csv'

classes = ['person', 'face', 'cellphone', 'hand']
data = []

for txt_file in os.listdir(txt_folder):
    if txt_file.endswith('.txt'):
        txt_file_path = os.path.join(txt_folder, txt_file)
        
        labels = {class_name: 0 for class_name in classes}

        with open(txt_file_path, 'r') as file:
            for line in file:
                parts = line.strip().split()

                class_id = int(parts[0])
                if class_id in class_mapping:
                    class_name = class_mapping[class_id]
                    labels[class_name] = 1

        filename = os.path.splitext(txt_file)[0] + '.jpg'
        image_path = os.path.join(txt_folder, filename)
        row = {'filename': image_path}
        row.update(labels)
        data.append(row)

df = pd.DataFrame(data)

df.to_csv(output_csv, index=False)
print(f"CSV file saved to {output_csv}")
```

```python
import os
import pandas as pd

csv_folder = 'data'
output_csv = 'data/all_merged_annotations.csv'

all_data = []

for csv_file in os.listdir(csv_folder):
    if csv_file.endswith('.csv'):
        file_path = os.path.join(csv_folder, csv_file)

        df = pd.read_csv(file_path)

        all_data.append(df)

combined_df = pd.concat(all_data, ignore_index=True)

# Save the combined DataFrame to a CSV
combined_df.to_csv(output_csv, index=False)

print(f"Combined CSV saved to {output_csv}")
```
